### Data Access & Setup

To run this notebook, the dataset **CardKrueger_1994_Fastfood.xlsx** is required.

The notebook automatically downloads and imports the dataset directly from our public GitHub repository. One can run all cells sequentially with no manual setup required.

### Load libraries, load dataset via GitHub

Run the cell below to fetch the dataset directly from our GitHub repository.

```python
!pip install pyfixest maketables -q
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pyfixest as pf
import maketables as mt

sns.set_theme(style="whitegrid")

# Direct raw link to the Excel file on GitHub
url_to_excel = "https://raw.githubusercontent.com/nhentze1/Minimum-Wages-and-Business-Models-Company-Owned-vs.-Franchise-Restaurants/main/CardKrueger_1994_Fastfood.xlsx"

try:
    # Load the Excel file from the GitHub URL
    excel_file = pd.ExcelFile(url_to_excel)
    print("Successfully loaded! Available sheets:", excel_file.sheet_names)

    # Load the main data sheet directly
    df = pd.read_excel(url_to_excel, sheet_name='Daten')
    print("\n--- First rows of the main data sheet ('Daten') ---")
    display(df.head())
except Exception as e:
    print("Error loading data from GitHub:", e)
```

### Show and load all sheets in the Excel file

```python
# Show the names of all sheets in the file
if 'excel_file' in globals():
    print("Available sheets in the file:", excel_file.sheet_names)
else:
    print("Please run the setup cells above first.")
```
